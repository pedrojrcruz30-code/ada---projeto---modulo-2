# 1. Progressão salarial por nível de experiência

````python
# --- Parte A: Salário médio por nível de experiência geral ---
print("--- Média Salarial por Nível de Experiência Geral ---")
# Agrupa por nível de experiência e calcula a média do salário em USD
media_por_experiencia = df.groupby('experience_level')['salary_in_usd'].mean().round(2).reset_index()
display(media_por_experiencia)

print("\n" + "="*50 + "\n")

# --- Parte B: Comparação entre os cargos principais ---
print("--- Progressão Salarial para os Principais Cargos ---")

# 1. Definimos a lista dos cargos que queremos filtrar
cargos_principais = ['Data Analyst', 'Data Scientist', 'Data Engineer']

# 2. Filtramos o DataFrame original para conter apenas esses cargos
df_filtrado_cargos = df[df['job_title'].isin(cargos_principais)]

# 3. Agrupamos por cargo E nível de experiência para ver a progressão de cada um
progressao_cargos = df_filtrado_cargos.groupby(['job_title', 'experience_level'])['salary_in_usd'].mean().unstack().round(2)

# Exibe o resultado em formato de tabela cruzada para facilitar a comparação visual
display(progressao_cargos)
````

# 2. Trabalho remoto compensa?

````python
# Parte A: Comparação Geral por Modelo de Trabalho ---
print("--- Média Salarial por Modelo de Trabalho ---")

media_por_modelo = df.groupby('work_models')['salary_in_usd'].mean().round(2).reset_index()
display(media_por_modelo)

display("="*150)

# Parte B: Modelo de Trabalho Cruzado com Nível de Experiência ---
print("--- Média Salarial por Modelo de Trabalho e Nível de Experiência ---")

# Agrupamos por ambas as colunas e calculamos a média
modelo_e_experiencia = df.groupby(['experience_level', 'work_models'])['salary_in_usd'].mean().unstack().round(2)
display(modelo_e_experiencia)
````

#3. **Onde estão as melhores oportunidades?**
###Quais países da residência do empregado têm os maiores salários médios? Existe correlação entre país e modalidade de trabalho (remoto vs presencial)?
